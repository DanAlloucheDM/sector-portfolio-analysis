```python
import pandas as pd
import matplotlib.pyplot as plt
from src.data.market_ import Market

print("\n=== TEST DE LA CLASSE MARKET ===\n")
```

```python
# 1. Initialisation 
print("1. Initialisation...")
market = Market(
    start_date='2014-01-01',
    end_date='2024-12-31',
    data_dir='data_test',
    verbose=True,
    limit=2,  # Seulement 2 instruments par secteur et type
    save=True,
    rounding=True
)
```

```python
# 2. Test des méthodes d'accès aux données
print("\n2. Test des méthodes d'accès aux données...")

# Afficher les secteurs disponibles
print(f"\nSecteurs disponibles: {market.sectors}")

# Récupérer les tickers par secteur (exemple avec technologie)
tech_tickers = market.get_tickers_by_sector('technology')
print(f"\nTickers du secteur technologique: {tech_tickers}")
```

```python
# Récupérer les données d'un secteur
if tech_tickers and tech_tickers.get('stocks'):
    tech_data = market.get_sector_data('technology')
    print(f"\nAperçu des données du secteur technologique:")
    print(tech_data.head())
    
    # Récupérer les données d'un ticker spécifique
    ticker = tech_tickers['stocks'][0] if tech_tickers['stocks'] else None
    if ticker:
        ticker_data = market.get_ticker_data(ticker)
        print(f"\nAperçu des données pour {ticker}:")
        print(ticker_data.head())
```

```python
# 3. Test des méthodes d'analyse des dates manquantes
print("\n3. Test des méthodes d'analyse des dates manquantes...")

# Obtenir les informations sur les tickers
tickers_info = market.get_ticker_info()
print(f"\nAperçu des informations sur les tickers:")
print(tickers_info.head())
```

```python
# Identifier les tickers avec des dates manquantes
missing_tickers = market.get_tickers_with_missing_dates(sort_by='percentage')
print(f"\nTickers avec dates manquantes:")
if not missing_tickers.empty:
    # Afficher les informations principales avec formatage amélioré
    print(missing_tickers[['Ticker', 'Sector', 'Type', 'NombreDatesMissing', 'PourcentageMissing']].to_string(
        formatters={
            'PourcentageMissing': '{:.2f}%'.format
        },
        index=False
    ))
    
    # Afficher quelques exemples de dates manquantes pour un ticker
    if len(missing_tickers) > 0:
        example_ticker = missing_tickers.iloc[0]['Ticker']
        missing_dates = missing_tickers.iloc[0]['MissingDates']
        if missing_dates and len(missing_dates) > 0:
            print(f"\nExemples de dates manquantes pour {example_ticker}:")
            dates_to_show = min(5, len(missing_dates))
            for date in sorted(missing_dates)[:dates_to_show]:
                print(f"  - {date.strftime('%Y-%m-%d')}")
            if len(missing_dates) > 5:
                print(f"  ... et {len(missing_dates) - 5} autres dates")
else:
    print("Aucun ticker ne présente de dates manquantes.")
```

```python
# 4. Test des méthodes de remplissage des données manquantes
print("\n4. Test des différentes méthodes de remplissage...")

if not missing_tickers.empty:
    # Sélectionner le premier ticker avec des dates manquantes pour les tests
    test_ticker = missing_tickers.iloc[0]['Ticker']
    print(f"\nTest avec le ticker: {test_ticker}")
    
    # Données avant remplissage
    before_data = market.get_ticker_data(test_ticker)
    print(f"Nombre de lignes avant remplissage: {len(before_data)}")
    
    # Obtenir une date manquante pour visualisation
    if missing_tickers.iloc[0]['MissingDates']:
        missing_date = sorted(missing_tickers.iloc[0]['MissingDates'])[0]
        print(f"\nExemple de données autour de la date manquante {missing_date.strftime('%Y-%m-%d')}:")
        
        # Trouver des dates voisines pour l'affichage (3 jours avant et après)
        dates_index = pd.to_datetime(before_data.index.get_level_values('Date'))
        
        # Trouver 3 dates avant et après la date manquante qui sont dans le jeu de données
        dates_before = [d for d in dates_index if d < missing_date][-3:] if len([d for d in dates_index if d < missing_date]) >= 3 else [d for d in dates_index if d < missing_date]
        dates_after = [d for d in dates_index if d > missing_date][:3] if len([d for d in dates_index if d > missing_date]) >= 3 else [d for d in dates_index if d > missing_date]
        
        # Créer un masque pour ces dates
        date_mask = before_data.index.get_level_values('Date').isin(dates_before + dates_after)
        context_data = before_data[date_mask].copy()
        
        # Réinitialiser l'index pour un affichage plus clair
        display_data = context_data.reset_index()
        display_data = display_data[['Date', 'Close', 'Open', 'High', 'Low', 'Volume']]
        
        # Mettre en évidence où la date manquante devrait être
        print("Données avant la date manquante:")
        if dates_before:
            print(display_data[display_data['Date'].isin(dates_before)].sort_values('Date'))
        else:
            print("Aucune donnée disponible avant la date manquante")
            
        print(f"\n>>> DATE MANQUANTE: {missing_date.strftime('%Y-%m-%d')} <<<\n")
        
        print("Données après la date manquante:")
        if dates_after:
            print(display_data[display_data['Date'].isin(dates_after)].sort_values('Date'))
        else:
            print("Aucune donnée disponible après la date manquante")
    
    # a. Test du remplissage avec la méthode "previous"
    print("\na. Méthode 'previous' (valeur précédente):")
    # Créer une copie du marché original pour restaurer après le test
    market_original = market.market.copy()
    tickers_info_original = market.tickers_info.copy()
    
    # Appliquer la méthode de remplissage
    result = market.fill_missing_dates(method='previous', min_missing=1)
    
    # Vérifier les résultats
    after_data = market.get_ticker_data(test_ticker)
    print(f"Nombre de lignes après remplissage: {len(after_data)}")
    print(f"Résultat de l'opération: {'Succès' if result else 'Échec'}")
    
    # Afficher les données après remplissage autour de la date manquante
    if missing_tickers.iloc[0]['MissingDates']:
        print(f"\nDonnées après remplissage 'previous' pour la date {missing_date.strftime('%Y-%m-%d')}:")
        
        # Inclure la date précédemment manquante
        dates_to_show = dates_before + [missing_date] + dates_after
        date_mask = after_data.index.get_level_values('Date').isin(dates_to_show)
        filled_data = after_data[date_mask].copy().reset_index()
        filled_data = filled_data[['Date', 'Close', 'Open', 'High', 'Low', 'Volume']]
        print(filled_data.sort_values('Date'))
        
        # Mettre en évidence la ligne ajoutée
        filled_row = filled_data[filled_data['Date'] == missing_date]
        if not filled_row.empty:
            print("\nValeurs utilisées pour la date manquante:")
            print(filled_row)
    
    # Restaurer le marché original
    market.market = market_original.copy()
    market.tickers_info = tickers_info_original.copy()
    
    # b. Test du remplissage avec la méthode "next"
    print("\nb. Méthode 'next' (valeur suivante):")
    result = market.fill_missing_dates(method='next', min_missing=1)
    
    # Vérifier les résultats
    after_data = market.get_ticker_data(test_ticker)
    print(f"Nombre de lignes après remplissage: {len(after_data)}")
    print(f"Résultat de l'opération: {'Succès' if result else 'Échec'}")
    
    # Afficher les données après remplissage autour de la date manquante
    if missing_tickers.iloc[0]['MissingDates']:
        print(f"\nDonnées après remplissage 'next' pour la date {missing_date.strftime('%Y-%m-%d')}:")
        dates_to_show = dates_before + [missing_date] + dates_after
        date_mask = after_data.index.get_level_values('Date').isin(dates_to_show)
        filled_data = after_data[date_mask].copy().reset_index()
        filled_data = filled_data[['Date', 'Close', 'Open', 'High', 'Low', 'Volume']]
        print(filled_data.sort_values('Date'))
        
        # Mettre en évidence la ligne ajoutée
        filled_row = filled_data[filled_data['Date'] == missing_date]
        if not filled_row.empty:
            print("\nValeurs utilisées pour la date manquante:")
            print(filled_row)
    
    # Restaurer le marché original
    market.market = market_original.copy()
    market.tickers_info = tickers_info_original.copy()
    
    # c. Test du remplissage avec la méthode "interpolate"
    print("\nc. Méthode 'interpolate' (interpolation linéaire):")
    result = market.fill_missing_dates(method='interpolate', min_missing=1)
    
    # Vérifier les résultats
    after_data = market.get_ticker_data(test_ticker)
    print(f"Nombre de lignes après remplissage: {len(after_data)}")
    print(f"Résultat de l'opération: {'Succès' if result else 'Échec'}")
    
    # Afficher les données après remplissage autour de la date manquante
    if missing_tickers.iloc[0]['MissingDates']:
        print(f"\nDonnées après remplissage 'interpolate' pour la date {missing_date.strftime('%Y-%m-%d')}:")
        dates_to_show = dates_before + [missing_date] + dates_after
        date_mask = after_data.index.get_level_values('Date').isin(dates_to_show)
        filled_data = after_data[date_mask].copy().reset_index()
        filled_data = filled_data[['Date', 'Close', 'Open', 'High', 'Low', 'Volume']]
        print(filled_data.sort_values('Date'))
        
        # Mettre en évidence la ligne ajoutée
        filled_row = filled_data[filled_data['Date'] == missing_date]
        if not filled_row.empty:
            print("\nValeurs interpolées pour la date manquante:")
            print(filled_row)
    
    # Restaurer le marché original
    market.market = market_original.copy()
    market.tickers_info = tickers_info_original.copy()
    
    # d. Test du remplissage avec la méthode "mean"
    print("\nd. Méthode 'mean' (moyenne):")
    result = market.fill_missing_dates(method='mean', min_missing=1)
    
    # Vérifier les résultats
    after_data = market.get_ticker_data(test_ticker)
    print(f"Nombre de lignes après remplissage: {len(after_data)}")
    print(f"Résultat de l'opération: {'Succès' if result else 'Échec'}")
    
    # Afficher les données après remplissage autour de la date manquante
    if missing_tickers.iloc[0]['MissingDates']:
        print(f"\nDonnées après remplissage 'mean' pour la date {missing_date.strftime('%Y-%m-%d')}:")
        dates_to_show = dates_before + [missing_date] + dates_after
        date_mask = after_data.index.get_level_values('Date').isin(dates_to_show)
        filled_data = after_data[date_mask].copy().reset_index()
        filled_data = filled_data[['Date', 'Close', 'Open', 'High', 'Low', 'Volume']]
        print(filled_data.sort_values('Date'))
        
        # Mettre en évidence la ligne ajoutée
        filled_row = filled_data[filled_data['Date'] == missing_date]
        if not filled_row.empty:
            print("\nValeurs moyennes pour la date manquante:")
            print(filled_row)
    
    # Vérifier les nouvelles dates manquantes après remplissage complet
    missing_after = market.get_tickers_with_missing_dates(min_missing=1)
    print(f"\nTickers avec dates manquantes après remplissage final:")
    if not missing_after.empty:
        print(missing_after[['Ticker', 'NombreDatesMissing']])
    else:
        print("Aucune date manquante après remplissage!")
else:
    print("Aucun ticker avec dates manquantes pour tester le remplissage.")
```
